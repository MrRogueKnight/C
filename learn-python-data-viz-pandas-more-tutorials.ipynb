**Arithmetic**

->In general, Python follows the PEMDAS rule when deciding the order of operations.

->Parentheses, Exponents, Multiplication/Division (from left to right), Addition/Subtraction (from left to right)

You can actually do a lot of calculations with python! See the table below for some examples.
1. Addition 	(+)
2. Subtraction 	(-) 	
3. Multiplication 	(*) 	
4. Division 	(/) 	  
5. Exponent 	(**) 	  

```python
# Verify the answer
print(10+2-3*4/2**2) 
```

```python
print(-10+2-(3*(4/(2**2)))) #You can control the order of operations in long calculations with parentheses.
```

**Variables** 
1. can't have spaces
2. includes letters, numbers, and underscores this means special charachers are not allowed
3. starts with a letter or underscore not numbers

```python
# Create a variable called test_var and give it a value of 20 ** 2
test_var = 20 ** 2

# Print the value of test_var
print(test_var)
```

```python

# Updating a variable 

my_var = 10 # Set the value 
print(my_var)

my_var = my_var + 3 # Update/Change the value (We can directly assign or even perform PEDMAS)
print(my_var)

```
